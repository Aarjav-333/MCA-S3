### Q01. Element-wise array comparisons

**Question:** Write a NumPy program to compare two arrays element by element using >, <, >=, <= and ==.

### Q20. Complete matrix analysis

**Question:** Accept two user-defined matrices. Find product, transpose, determinant, inverse, trace, rank and eigenvalues/eigenvectors wherever defined.

**Question type:** Notebook-based exercise.

**Answer (Python code):**

```python
import numpy as np
from unittest.mock import patch

def input_matrix(label):
    rows, cols = map(int, input(f'{label} dimensions: ').split())
    values = list(map(float, input(f'{label} values: ').split()))
    if len(values) != rows * cols:
        raise ValueError('Incorrect number of matrix elements')
    return np.array(values).reshape(rows, cols)

def report(a, b):
    print('A:\n', a, '\nB:\n', b)
    if a.shape[1] == b.shape[0]:
        print('A @ B:\n', a @ b)
    else:
        print('A @ B: incompatible dimensions')
    for label, matrix in [('A', a), ('B', b)]:
        print(f'\n{label} transpose:\n', matrix.T)
        print(f'{label} rank:', np.linalg.matrix_rank(matrix))
        if matrix.shape[0] != matrix.shape[1]:
            print(f'{label}: determinant, inverse, trace and eig require a square matrix here')
            continue
        det = np.linalg.det(matrix)
        print(f'{label} determinant:', round(det, 4), 'trace:', np.trace(matrix))
        if not np.isclose(det, 0):
            print(f'{label} inverse:\n', np.linalg.inv(matrix))
        else:
            print(f'{label} inverse: singular matrix')
        eigvals, eigvecs = np.linalg.eig(matrix)
        print(f'{label} eigenvalues:', np.round(eigvals, 4))
        print(f'{label} eigenvectors:\n', np.round(eigvecs, 4))

with patch('builtins.input', side_effect=['2 2', '1 2 3 4', '2 2', '2 0 0 3']):
    a = input_matrix('A')
    b = input_matrix('B')
report(a, b)
```

**Explanation:** The calculations are defined only under appropriate conditions. Matrix multiplication needs matching inner dimensions; inverse needs an invertible square matrix. The code prevents incompatible operations and uses simulated keyboard inputs for the saved example.

**Output:** The executed code cell immediately above contains its saved text output and/or graph. Run the cell again to regenerate it.

**Explanation:** A square matrix is singular when its determinant is zero. `np.isclose` is a practical guard against tiny floating-point errors.

### Q49. KNN on the Diabetes dataset

**Question:** Convert the continuous Diabetes target into three bins using mean ± standard deviation, scale features, train KNN with k=5 and calculate accuracy.

**Question type:** Notebook-based exercise.

**Answer (Python code):**

```python
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
db = load_diabetes()
X, y = db.data, db.target
bins = [y.mean() - y.std(), y.mean() + y.std()]
y_class = np.digitize(y, bins=bins)  # classes 0, 1, 2
X_train, X_test, y_train, y_test = train_test_split(
    X, y_class, train_size=0.7, random_state=42, stratify=y_class)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
model = KNeighborsClassifier(n_neighbors=5).fit(X_train, y_train)
pred = model.predict(X_test)
print('Class counts:', np.bincount(y_class))
print('Accuracy:', round(accuracy_score(y_test, pred), 4))
```

**Explanation:** `np.digitize` bins the regression target into three categories for classification (as in the source). Standardization is fit **only on training data**, avoiding data leakage from the held-out test set.

### Q67. Calculate MSE and R² manually

**Question:** From arrays of actual and predicted numbers, calculate MSE and R² without model metric functions.

**Question type:** Related practice variation.

**Answer (Python code):**

```python
import numpy as np
actual = np.array([10., 20., 30.])
predicted = np.array([12., 18., 29.])
errors = actual - predicted
mse = np.mean(errors ** 2)
ss_res = np.sum(errors ** 2)
ss_total = np.sum((actual - actual.mean()) ** 2)
r2 = 1 - ss_res / ss_total
print('Residuals:', errors)
print('MSE:', round(mse, 4))
print('SS_res:', ss_res, 'SS_total:', ss_total)
print('R²:', round(r2, 4))
```

**Explanation:** MSE = mean((actual−predicted)²). R² = 1−SS_res/SS_total. R² can be negative; it equals 0 for a predictor that always returns the test-target mean.

### Q73. Effect of SVM C

**Question:** Train linear SVM classifiers with different C values and compare training and testing accuracy.

**Question type:** Related practice variation.

**Answer (Python code):**

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.svm import SVC
X, y = make_classification(n_samples=400, n_features=5, n_informative=3,
                           n_redundant=0, class_sep=0.7, flip_y=0.06, random_state=42)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
for C in [0.01, 0.1, 1, 10]:
    model = make_pipeline(StandardScaler(), SVC(kernel='linear', C=C))
    model.fit(Xtr, ytr)
    print(f'C={C:<5}: train accuracy={model.score(Xtr, ytr):.3f}, '
          f'test accuracy={model.score(Xte, yte):.3f}')
```

**Explanation:** C controls the penalty for margin violations: smaller C permits a softer margin; larger C penalizes errors more strongly. Better training fit does not guarantee better test accuracy.
